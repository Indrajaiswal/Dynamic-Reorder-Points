```python
product_counts = df.groupby(
    ["Product_ID", "Product_Name", "Category"]
).size().reset_index(name="Record_Count")

product_counts
```

```python
product_counts.sort_values("Record_Count")
```

```python
df["Date"] = pd.to_datetime(df["Date"])

yearly_product_counts = (
    df.groupby([df["Date"].dt.year, "Product_Name"])
      .size()
      .unstack(fill_value=0)
)

yearly_product_counts
```
