```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("../data/processed/feature_engineered_data.csv")

df["Date"] = pd.to_datetime(df["Date"])

print("Dataset shape:", df.shape)

df.head()
```

```python
print("Columns:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDate range:")
print("Start:", df["Date"].min())
print("End:", df["Date"].max())
```

```python
df = df.sort_values(["Date", "Product_ID"]).reset_index(drop=True)

df.head()
```

```python
feature_columns = [
    "Product_ID",
    "Category",
    "Brand",
    "Current_Stock",
    "Unit_Price",
    "Discount_Percent",
    "Promotion",
    "Holiday",
    "Lead_Time_Days",
    "Safety_Stock",
    "Stockout",
    "Year",
    "Month",
    "Day",
    "Day_of_Week",
    "Week_of_Year",
    "Is_Weekend",
    "Lag_1_Day",
    "Lag_7_Days",
    "Lag_14_Days",
    "Rolling_7_Day",
    "Rolling_30_Day",
    "Rolling_7_Day_Std",
    "Discount_Amount",
    "Final_Price"
]

X = df[feature_columns]
y = df["Units_Sold"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
print("Target statistics:")
print(y.describe())
```

```python
print("\nUnique sales values:", y.nunique())
print("Minimum sales:", y.min())
print("Maximum sales:", y.max())
```

```python
unique_dates = sorted(df["Date"].unique())

split_index = int(len(unique_dates) * 0.8)

train_end_date = unique_dates[split_index]

train_df = df[df["Date"] < train_end_date].copy()
test_df = df[df["Date"] >= train_end_date].copy()

print("Training period:")
print(train_df["Date"].min(), "to", train_df["Date"].max())

print("\nTesting period:")
print(test_df["Date"].min(), "to", test_df["Date"].max())

print("\nTraining shape:", train_df.shape)
print("Testing shape:", test_df.shape)
```

```python
X_train = train_df[feature_columns]
y_train = train_df["Units_Sold"]

X_test = test_df[feature_columns]
y_test = test_df["Units_Sold"]

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
categorical_features = [
    "Product_ID",
    "Category",
    "Brand"
]

numerical_features = [
    "Current_Stock",
    "Unit_Price",
    "Discount_Percent",
    "Promotion",
    "Holiday",
    "Lead_Time_Days",
    "Safety_Stock",
    "Stockout",
    "Year",
    "Month",
    "Day",
    "Day_of_Week",
    "Week_of_Year",
    "Is_Weekend",
    "Lag_1_Day",
    "Lag_7_Days",
    "Lag_14_Days",
    "Rolling_7_Day",
    "Rolling_30_Day",
    "Rolling_7_Day_Std",
    "Discount_Amount",
    "Final_Price"
]
```

```python
print("Categorical features:", categorical_features)
print("Numerical features:", numerical_features)
print("Total features:", len(categorical_features) + len(numerical_features))
```

```python
print(len(feature_columns))
print(feature_columns)
```

preprocessing

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)

X_test_processed = preprocessor.transform(X_test)

print("Processed X_train shape:", X_train_processed.shape)
print("Processed X_test shape:", X_test_processed.shape)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
def calculate_mape(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    
    mask = y_true != 0
    
    return np.mean(
        np.abs((y_true[mask] - y_pred[mask]) / y_true[mask])
    ) * 100
```

```python
def evaluate_model(model_name, y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mape = calculate_mape(y_true, y_pred)

    return {
        "Model": model_name,
        "MAE": mae,
        "RMSE": rmse,
        "MAPE (%)": mape
    }
```

```python
from sklearn.linear_model import LinearRegression

linear_model = LinearRegression()

linear_model.fit(
    X_train_processed,
    y_train
)
```

```python
linear_predictions = linear_model.predict(X_test_processed)
```

```python
linear_result = evaluate_model(
    "Linear Regression",
    y_test,
    linear_predictions
)

linear_result
```

```python
comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": linear_predictions
})

comparison.head(20)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=200,
    max_depth=15,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(
    X_train_processed,
    y_train
)
```

```python
rf_predictions = rf_model.predict(X_test_processed)
```

```python
rf_result = evaluate_model(
    "Random Forest",
    y_test,
    rf_predictions
)

rf_result
```

```python
results = pd.DataFrame([
    linear_result,
    rf_result
])

results
```
